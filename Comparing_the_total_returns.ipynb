```python
#first company
name1=input("Please enter your first name: ")
p01=float(input("Please enter your initial pricefor the first company?: "))
p11=float(input("Please enter your final price for the first company?: "))
div1=float(input("Please enter your the dividend for the first company?: "))
```

```python
#second company
name2=input("Please enter your second name:")
p02=float(input("Please enter your initial pricefor the second company?: "))
p12=float(input("Please enter your final price for the second company?: "))
div2=float(input("Please enter your the dividend for the second company?: "))
```

```python
# return and dividend for the first company
p_ret1= p11/p01 -1
p_ret1

div_y1= div1/p01
div_y1

total_ret1= p_ret1*div_y1
total_ret1
```

```python
# return and dividend for the second company
p_ret2= p12/p02 -1
p_ret2

div_y2= div2/p02
div_y2

total_ret2= p_ret2*div_y2
total_ret2
```

```python
# comparison
#print(p_ret, div_y, total_ret, sep=", ")
print(f"for {name1}, price return is {p_ret1 : .2f}. Your dividend yield is {div_y1}. Total return is {total_ret1}.")
print(f"for {name2}, price return is {p_ret2}. Your dividend yield is {div_y2}. Total return is {total_ret2}.")

if total_ret1 > total_ret2:
  print(f"{name1} is more profitable.")
else:
  print(f"{name2} is more profitable.")
```
